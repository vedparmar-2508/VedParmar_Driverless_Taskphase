```python
import numpy as np

def transform_global_to_local(cones, pose):
    x_c, y_c, theta_deg = pose
    theta = np.radians(theta_deg)
    c, s = np.cos(theta), np.sin(theta)
    
    # Rotation matrix from global to car frame
    R = np.array([[c, s], [-s, c]])
    
    local_cones = {}
    for name, (x_g, y_g) in cones.items():
        translated = np.array([x_g - x_c, y_g - y_c])
        local_cones[name] = R @ translated
        
    return local_cones

# Given inputs
initial_pose = (10.0, 5.0, 30.0)
v0 = 5.0
a = 1.0
omega_deg = 10.0
dt = 0.2

cones_global = {
    "B1": (12.0, 7.0), "B2": (8.0, 6.0),
    "Y1": (13.0, 4.0), "Y2": (9.0, 3.0)
}

# (i) Compute new pose after 0.2 s
dist = v0 * dt + 0.5 * a * (dt ** 2)
theta_new_deg = initial_pose[2] + omega_deg * dt
theta_avg = np.radians((initial_pose[2] + theta_new_deg) / 2.0)

x_new = initial_pose[0] + dist * np.cos(theta_avg)
y_new = initial_pose[1] + dist * np.sin(theta_avg)
new_pose = (x_new, y_new, theta_new_deg)

print(f"New Pose: x={x_new:.3f}, y={y_new:.3f}, theta={theta_new_deg:.1f} deg")

# (ii) Convert cone positions
local_initial = transform_global_to_local(cones_global, initial_pose)
local_new = transform_global_to_local(cones_global, new_pose)

print("\nInitial Local Cones:", {k: np.round(v, 3) for k, v in local_initial.items()})
print("New Local Cones    :", {k: np.round(v, 3) for k, v in local_new.items()})
```
